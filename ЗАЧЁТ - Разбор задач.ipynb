## Подбрасывание монетки

Классичный эксперимент в тервере, при котором, чем больше число его исходов, тем ближе вероятность выпадения орла или решки будет приближаться "50 на 50". Мы говорим, что вероятность исхода «решка» равна $0.5$.

В общем случае эксперимент — мы будем называть эксперименты измерениями — имеет ограниченное число возможных исходов. Обозначим их как $E_1, E_2, …, E_n$. Также примем, что результатом эксперимента, или измерения, может быть только один из этих исходов.

Выпадение результата $E_i$ имеет вероятность $P_i$. Вероятности — это числа от $0$ до $1$ и сумма вероятностей всех исходов равна $1$. В случае с броском монеты возможны два исхода, орел или решка. Если монета правильная, без изъянов, вероят
ность каждого события равна $\frac12$.

## [SLOP] Шифрование

**Шифрование** — это процесс преобразования открытого текста (plaintext) в зашифрованный текст (ciphertext) с использованием алгоритма и ключа. Цель шифрования — защитить информацию от несанкционированного доступа.

**Основные компоненты:**
- Открытый текст (M) — исходное сообщение
- Ключ шифрования (K) — секретная информация для шифрования
- Алгоритм шифрования (E) — математическая функция
- Шифротекст (C) — результат шифрования: C = E(M, K)
- Расшифрование (D) — обратная операция: M = D(C, K)

**Пример (шифр Цезаря aka сдвиг на N букв — один из старейших шифров):**


````python
def caesar_encrypt(text, shift):
    result = ""
    for char in text:
        if char.isalpha():
            start = ord('А') if char.isupper() else ord('а')
            result += chr((ord(char) - start + shift) % 32 + start)
        else:
            result += char
    return result

def caesar_decrypt(text, shift):
    return caesar_encrypt(text, -shift)

# Пример использования
message = "ПРИВЕТ"
encrypted = caesar_encrypt(message, 3)
decrypted = caesar_decrypt(encrypted, 3)

print(f"Исходное сообщение: {message}")
print(f"Зашифрованное: {encrypted}")
print(f"Расшифрованное: {decrypted}")
````

## [SLOP] Сохранение сообщения в тайне

**Конфиденциальность** — основное свойство криптографических систем. Для обеспечения тайны сообщения необходимы:

- Стойкий алгоритм — даже зная алгоритм, без ключа невозможно расшифровать
- Достаточная длина ключа — чтобы перебор был вычислительно невозможен
- Случайность ключа — ключ не должен быть предсказуемым
- Защита от статистического анализа — шифротекст не должен содержать закономерностей

## [SLOP] Сохранение ключей шифрования

**Управление ключами** — критически важная задача в криптографии. Ключи должны быть:

- Генерированы случайно — с использованием криптографически стойких генераторов
- Храниться безопасно — в зашифрованном виде, в аппаратных модулях (HSM)
- Передаваться по защищённым каналам — с использованием протоколов обмена ключами
Уничтожаться после использования — для одноразовых ключей

**Методы защиты ключей:**

- Аппаратные токены и смарт-карты
- Разделение ключа между несколькими лицами (secret sharing)
- Использование мастер-ключей для шифрования других ключей
- Ограничение времени жизни ключа

## Жизненный цикл ключей шифрования

1. **Регистрация пользователей** (удостовериться, что вы — это вы)
2. **Инициализация**
3. **Генерация ключей** (создание нового ключа шифрования)
4. **Установка ключей** (связывание этого ключа с вами, в случае физического контакта буквально выдача его вам лично на руки)
5. **Регистрация ключей** (внесение этих связанных ключей в базу данных)
6. **Обычный режим работы**
7. **Хранение ключа** (например, в локальном центре сертификации компании)
8. **Замена ключа** (в случае, если он был скомпрометирован или, более простая ситуация, истёк его срок жизни — такой ключ архивируется)
9. **Архивирование** — Внесение в базу истёкших ключей. Т.е. он по-прежнему работает для уже зашифрованных им файлов, но для новых файлов мы его больше не используем, чтобы усложнить злоумышленнику подбор данных. ЕСЛИ КЛЮЧ СКОМПРОМЕТИРОВАН, КЛЮЧ НЕ АРХИВИРУЕТСЯ — вместо этого происходит отмена ключа
10. **Уничтожение ключей**
11. **Восстановление ключей**
12. **Отмена ключей** — ВСЕ зашифрованные этим ключом данные расшифровываются и зашифровываются каким-то новым ключом, после чего старый ключ уничтожается.

## Принцип блокнота

1. Длина ключа = длина сообщения.

2. Каждый ключ должен использоваться строго 1 раз (чтобы не допустить статистического анализа).

Свойства:
- Ключ — полностью случайный.
- При соблюдении условий, абсолютная информационно-теоретическая стойкость, т.е. невозможно вскрыть даже с бесконечной вычислительной мощностью.

Проблемы:
- Нереалистично хранить и передавать ключи такой длины.
- Повторное использование ключа полностью ломает безопасность.

## Симметричное шифрование

Если для шифровки и дешифровки используется один и тот же ключ, такой шифр назызвается **симметричным**. Такие шифры используются, например, в ранних протоколах VPN:
$$C=E_k(m), \ m = D_k(C)$$
Где: $C$ — шифротекст, $m$ — исходное сообщение, $k$ — ключ,
$E$ и $D$ — функции расшифровки и дешифровки соответственно

Чем труднее этот самый ключ, тем лучше. Например, трудные математические задачи:
$$y = x^e \ mod \ m, \ x = y^d \ mod \ m$$

## [SLOP] Cимметричное шифрование - Недостатки

1. **Перехват ключа «по середине» (атака Man In The Middle, MITM)** — владелец ключа, или просто перехвативший ключ злоумышленник, может читать ВООБЩЕ ВСЁ и даже подменять данные.

2. **Проблема передачи ключа** — ключ нужно как-то безопасно передать: по незащищённому каналу опасно, а защищённый канал уже требует какого-то другого защищённого способа передачи.

Это логический тупик, который долгое время был основной проблемой криптографии. Единственным его решением является АБСОЛЮТНОЕ доверие каналу передачи (доверенные сертификаты, например).

## Асимметричное шифрование - Передача ключей

В асимметричной криптографии используется пара ключей — **открытый (public)** и **закрытый (private)**.

Открытый ключ можно передавать по открытому каналу, но закрытый ключ не передаётся никогда.

## Асимметричное шифрование - Авторизация

Типичный сценарий выглядит как-то так:

1. Клиент генерирует случайный симметричный ключ (например, на один сеанс)

2. Перед подключением к серверу, клиент шифрует свой закрытый ключ открытым ключом сервера

3. Клиент передаёт ключ серверу, после чего тот расшифровывает его своим закрытым ключом

4. Обе стороны используют открытый ключ для передачи данных уже по протоколу симметричного шифрования (т.е. данные шифруются закрытым ключом клиента в оба направления)

Таким образом, проблемы передачи ключа нет, но полностью симметрию асимметрия не заменяет.

## [SLOP] Асиммеричное шифрование - Недостатки

**Слишком медленные**

- Операции с большими числами (2048–4096 бит) очень долгие. Поэтому асимметричные шифры обычно используют только для обмена ключами и аутентификации (например, в SSH и HTTPS).

**Подвержены квантовым атакам**

- Алгоритм Шора может взломать RSA, DSA, ECC.

- Алгоритм Гловела ослабляет симметричные шифры (в 2 раза по длине ключа).

**Следствие**

- Симметричные шифры относительно устойчивы (нужно просто увеличить ключ),

- Асимметричные — фундаментально уязвимы.

## RSA (открытые ключи)

RSA — асимметричный алгоритм шифрования, основанный на сложности факторизации больших чисел.

1. Возьмем два больших простых числа $p$ и $q$.
2. Определим n, как результат умножения $p$ на $q$: $n=p*q$.
3. Выберем случайное число, которое назовем $d$. **Это число должно быть
взаимно простым (не иметь ни одного общего делителя, кроме $1$) с
результатом умножения** $(p-1)*(q-1)$.
4. Определим такое число $е$, для которого является истинным
следующее соотношение $(e*d) \mod ((p-1)*(q-1))=1$.
5. Hазовем **открытым ключом** числа $e$ и $n$, а **секретным (закрытым)** - $d$ и $n$.

### Пример без кода ололо:

1. Выберем $p=3$ и $q=11$.
2. Определим $n= 3*11=33$.
3. Hайдем $(p-1)*(q-1)=20$.

   Следовательно, d будет равно, например, 3.
4. Выберем число е по следующей формуле:
$(e*3) \mod 20=1$.

   Значит е будет равно, например, 7.
5. Представим шифруемое текстовое сообщение на русском языке как последовательность чисел в диапозоне от 0 (=А) до 32 (=Я).
6. Зашифруем сообщение, используя открытый
ключ $\{7,33\}$:
$$
C1 = (3^7) \mod 33 = 2187 \mod 33 = 9 \\
C2 = (1^7) \mod 33 = 1 \mod 33 = 1 \\
C3 = (2^7) \mod 33 = 128 \mod 33 = 29 \\
$$

7. Теперь расшифруем данные, используя закрытый
ключ $\{3,33\}$:
$$
M1=(9^3) \mod 33 =729 \mod 33 = 3 \ (=В) \\
M2=(1^3) \mod 33 =1 \mod 33 = 1 \ (=А) \\
M3=(29^3) \mod 33 = 24389 \mod 33 = 2 \ (=Б) \\
$$

Ещё одна шняга, основанная на RSA.

**Суть — Алиса-пиписа доказывает Бобу, что у неё есть закрытый ключ $d$ (т.е. что она является легитимным центром авторизации), при этом сам ключ не показывая.**

Лолгоритм:

1. Бiб даёт Алисе челлендж ~~24 часа ебаться в чугуне с волосатыми мужиками обмазанными маслом~~. Он отправляет ей случайное сообщение $s$.

2. Алиса, получившая сообщение ~~по хорошему должна свалить, ибо даже я охуел бы от таких предложений~~, расшифровывает его своим закрытым ключом: $g = s^d % m$. По сути она оставляет свою подпись на полученном сообщении, чтобы подтвердить, что именно она его получила.

3. Бiб проверяет, было ли подписано сообщение настоящим закрытым ключом при помощи открытого ключа $e$: $s’ = g^e % m$.

Если $s’ = s$, значит всё заебись, чётко, это ответила именно Алиса-пиписа!

**Почему это работает**

- Без знания $d$ невозможно вычислить корректный $g$

- Злоумышленник не может подделать подпись

- Это стандартная RSA-аутентификация

### Хуцпа с лекции (easier than you thought!):

image.png

````python
class AuthenticationProtocol:
    def __init__(self, p=61, q=53, e=17):
        self.p = p
        self.q = q
        self.e = e

        self.n = p * q
        phi = (p - 1) * (q - 1)
        self.d = pow(self.e, -1, phi)

    def get_public_key(self):
        return (self.e, self.n)

    def generate_challenge(self):
        return random.randint(100, self.n - 1)

    def sign_challenge(self, challenge):
        return pow(challenge, self.d, self.n)

    def verify_response(self, challenge, response, public_key):
        e, n = public_key
        decrypted = pow(response, e, n)
        return decrypted == challenge

# Демонстрация протокола
print("=== Протокол аутентификации ===")

# Алиса создаёт свои ключи
alice = AuthenticationProtocol()
alice_public = alice.get_public_key()
print(f"Открытый ключ Алисы: {alice_public}")

# Боб запускает аутентификацию
bob = AuthenticationProtocol()
challenge = bob.generate_challenge()
print(f"Боб отправляет челлендж: {challenge}")

# Алиса отвечает
response = alice.sign_challenge(challenge)
print(f"Алиса отправляет ответ: {response}")

# Боб проверяет
is_authenticated = bob.verify_response(challenge, response, alice_public)
print(f"Боб принял ответ: {is_authenticated}")

# Попытка атаки
print("\n=== Попытка подделки ===")
fake_response = random.randint(100, alice.n - 1)
is_fake_auth = bob.verify_response(challenge, fake_response, alice_public)
print(f"Боб принят подделку: {is_fake_auth}")
````

## Электронная цифровая подпись (ЭЦП)

~~ДЭЦП~~ Забавная шняга, которой шифруют текст. При шифровании у Алисы и Бiба есть две пары ключей - закрытый и открытый. Благодаря таким определённым парам достигается целостность, подлинность и невозможность подделки. Электронно-цифровая подпись работает с использованием метода RSA:


1. Алиса и Бiб вычисляют через RSA открытый и закрытый ключи
2. Алиса хэширует b каким-нибудь способом (любым на самом деле, какой на зачёте попадётся).
3. Алиса вычисляет подпись s через закрытый ключ d: s = b^d % m и отправляет сообщение и саму подпись Бiбу.
4. Бiб расшифровывает сообщение открытым ключом e.
5. Если расшифрованный текст равен оригинальному отправленному, значит подпись действительно принадлежит Алисе.


````python
# === 1. "Генерация ключей" (захардкожена для простоты) ===
# Маленькие простые числа (так делать НЕЛЬЗЯ в реальности)

p = 61
q = 53
n = p * q              # 3233
phi = (p - 1) * (q - 1)

e = 17                 # открытая экспонента
d = 2753               # закрытая экспонента (подобрана заранее)

public_key = (e, n)
private_key = (d, n)

# === 2. Простейший "хэш" сообщения ===
def simple_hash(message: str) -> int:
    return sum(ord(c) for c in message)

# === 3. Подпись сообщения ===
def sign(message: str, private_key):
    d, n = private_key
    h = simple_hash(message)
    s = pow(h, d, n)
    return s

# === 4. Проверка подписи ===
def verify(message: str, signature, public_key):
    e, n = public_key
    h = simple_hash(message)
    h_check = pow(signature, e, n)
    return h == h_check

# === 5. Демонстрация ===
message = "HELLO"

signature = sign(message, private_key)
print("Подпись:", signature)

is_valid = verify(message, signature, public_key)
print("Подпись верна?", is_valid)
````

4. Алиса и Бiб отправляют $A$ и $B$ друг другу.
5. Затем вычисляют секреты:
$$ K_a == B^a \mod p == g^{(a*b)} \mod p $$
$$ K_b == A^b \mod p == g^{(a*b)} \mod p $$

1. У нас есть 32-битное число, но для простаты экскремента мы просто возьмём два 4-битных блока в начале.
2. Берём ключ K, левый блок L и правый блок R. Все три переменные должны быть переведены в двоичный формат.
3. Складываем L и ключ K (причём именно складываем, а не ORём или XORим), а потом извлекаем остаток от деления числа на 2^32.
4. Затем в дело вступает на Г(Л)IСТ 28147-89. Переводим сумму L+K % 2^32 обратно в десятичную систему. Находим строку с номером, соответствующим нашей сумме (например если у нас сумма в десятичной системе - 12, то смотрим строку 12).
5. Пробегаемся по всем блокам (в нашем случае только по левой части L) и переводим значения всех блоков в соответствующие значения строки
6. Делаем сдвиг влево 11 раз.
7. Получившийся результат XORим с правой частью R.

На этом заканчивается раунд, а также моя кукуха

image.png

````python
# Игрушечный S-box (один столбец таблицы 16×8)
SBOX = [
    14,  # 0 -> 14
    4,   # 1 -> 4
    13,  # 2 -> 13
    1,   # 3 -> 1
    2,   # 4 -> 2
    15,  # 5 -> 15
    11,  # 6 -> 11
    8,   # 7 -> 8
    3,   # 8 -> 3
    10,  # 9 -> 10
    6,   # 10 -> 6
    12,  # 11 -> 12
    5,   # 12 -> 5
    9,   # 13 -> 9
    0,   # 14 -> 0
    7    # 15 -> 7
]

def feistel_round_8bit(L, R, K):
    """
    Один игрушечный раунд Feistel/ГОСТ для 8-битного блока:
    L, R, K — числа 0..15 (оставляем всё 4-битным ради простоты).
    Возвращает (L_new, R_new).
    """
    # 1) Сложение L и K по модулю 16 (игрушечный аналог мод 2^32)
    X = (L + K) % 16

    # 2) Прогоняем X через S-box
    Y = SBOX[X]  # тоже 0..15

    # 3) Циклический сдвиг влево на 2 бита в 4-битном пространстве
    # (игрушечный аналог сдвига на 11 бит в 32 битах)
    Y_shifted = ((Y << 2) & 0b1111) | (Y >> 2)

    # 4) XOR с правым полублоком
    R_new = R ^ Y_shifted

    # 5) Перестановка полублоков (классический Feistel)
    L_new = R

    return L_new, R_new

# Пример: L=5, R=10, K=7 (как разбирали)
L, R, K = 5, 10, 7
print(f"Вход:  L={L}, R={R}, K={K}")

L1, R1 = feistel_round_8bit(L, R, K)
print(f"После раунда: L'={L1}, R'={R1}")
````

### Хм, а мне клавдия вот так написала:

**ГОСТ 28147-89** — советский и российский стандарт симметричного шифрования, основанный на сети Фейстеля.

**Принцип работы сети Фейстеля:**

1. Блок данных делится на две половины: L (левая) и R (правая)

2. На каждом раунде:

    1. Правая часть обрабатывается функцией F с ключом раунда
    2. Результат XOR'ится с левой частью
    3. Половины меняются местами

3. После всех раундов половины объединяются

Преимущества:

- Шифрование и расшифрование используют одну структуру
- Функция F не обязана быть обратимой
- Высокая криптостойкость при достаточном числе раундов

````python
class FeistelNetwork:
    """Упрощённая реализация сети Фейстеля"""

    def __init__(self, rounds=8):
        self.rounds = rounds
        # Упрощённые ключи раундов (в реальности генерируются из мастер-ключа)
        self.round_keys = [i * 13 % 256 for i in range(rounds)]

    def f_function(self, right, key):
        """Функция раунда (упрощённая)"""
        # В реальном ГОСТ используются S-блоки и более сложные преобразования
        return ((right + key) * 7) % 256

    def split_block(self, block):
        """Разделение блока на две половины"""
        mid = len(block) // 2
        return block[:mid], block[mid:]

    def feistel_round(self, left, right, key):
        """Один раунд Фейстеля"""
        f_result = self.f_function(right, key)
        new_left = right
        new_right = left ^ f_result
        return new_left, new_right

    def encrypt_block(self, block_int):
        """Шифрование блока"""
        # Разделяем на левую и правую части
        left = (block_int >> 8) & 0xFF
        right = block_int & 0xFF

        # Раунды шифрования
        for i in range(self.rounds):
            left, right = self.feistel_round(left, right, self.round_keys[i])

        # Объединяем части (последний обмен не делается)
        return (right << 8) | left

    def decrypt_block(self, block_int):
        """Расшифрование блока (ключи в обратном порядке)"""
        left = (block_int >> 8) & 0xFF
        right = block_int & 0xFF

        for i in range(self.rounds - 1, -1, -1):
            left, right = self.feistel_round(left, right, self.round_keys[i])

        return (right << 8) | left

# Пример использования
feistel = FeistelNetwork(rounds=8)

plaintext = 0x1234  # 16-битный блок
print(f"Исходный текст: {hex(plaintext)}")

ciphertext = feistel.encrypt_block(plaintext)
print(f"Зашифрованный: {hex(ciphertext)}")

decrypted = feistel.decrypt_block(ciphertext)
print(f"Расшифрованный: {hex(decrypted)}")
print(f"Совпадает с оригиналом: {plaintext == decrypted}")
````

## [SLOP] Коллапс волновой функции

**Коллапс волновой функции** — фундаментальное явление квантовой механики: при измерении квантовой системы её состояние мгновенно переходит из суперпозиции в одно из базисных состояний.

Ключевые моменты:
- До измерения кубит находится в суперпозиции: $|ψ⟩ = α|0⟩ + β|1⟩$
- Вероятность получить $|0⟩$: $P(0) = |α|²$
- Вероятность получить $|1⟩$: $P(1) = |β|²$
- После измерения система находится в определённом состоянии
- Повторное измерение даст тот же самый результат, что получился при прошлом измерении

### **Первый закон квантовых джунглей**
Измерение изменяет систему.

Невозможно узнать состояние кубита, не разрушив суперпозицию.

````python
qc = QuantumCircuit(2)

qc.h(0)     # Вентиль Амидамару:
            # - перевод 0-го кубита в состояние неопределённости
            #   (оно же суперпозиция -- кубит равен и 1 и 0 одновременно!)

# Таким образом, мы достигаем состояния неопределённости 1-го кубита

qc.cx(0, 1) # Применяем C-NOT:
            # - если 0-й кубит = 1, применяем вентиль X (NOT) на 1-й кубит
            # - если 0-й кубит = 0, не применяем этот вентиль

# Таким образом, мы достигаем состояния запутанности 1-го и 2-го кубит

qc.z(1)     # Применяем сдвиг по фазе (поворот на π; |0> -> |0>, |1> -> -|1>)

qc.x(1)     # Применяем фильтр NOT (|0> -> |1>, |1> -> |0>)

qc.draw("mpl")
````

Стоит также помнить, что для состояний Белла справедливо:

$$ (X⊗I)∣Φ+⟩=(I⊗X)∣Φ+⟩ $$
$$ (Z⊗I)∣Φ+⟩=(I⊗Z)∣Φ+⟩ $$

То есть локальные операции на любом одном кубите переводят $Φ⁺$ в другие состояния Белла.

Это специфическое свойство максимальной запутанности.

## Эволюция системы

Просто идём по квантовой цепи по шагам (как если бы мы "нанизывали" вентили один за другим) и расписываем, как меняется значение всех кубитов на выходе от добавления к добавлению.

**Что важно помнить:**

- Порядок кубитов в векторе соответствует из порядку на обозначении. Например, в $|01>$ первый ~~фотон~~ кубит имеет значение $0$, а второй — $1$.

- В QISKIT кубиты перечисляются В ОБРАТНОМ ПОРЯДКЕ! То есть `110` означает $|011>$ и т.п.

- Помимо этого, все кубиты по умолчанию определены и равны `0`.

- `I` ничего не делает, вот прям вообще

- `X` меняет $0$ на $\pm1$, $\pm1$ на $0$

- `Z` меняет $1$ на $-1$, $-1$ на $1$, $0$ не меняет

- `H` меняет $0$ на $(|0> + |1>)$, $1$ на $(|0> - |1>)$

- `CX`, `CZ` работают аналогично `X` и `Z`, но кубит, где сам вентиль, срабатывает, только если кубит, от которого тянется линия ("контроллирующий") равен $1$ (иначе он не делает ничего)

- `CCX` (Тоффоли), `CCZ` — ну тут несложно догадаться, чем они отличаются от `CX` и `CZ`

**Пример:**

````python
qc = QuantumCircuit(2)

qc.h(0)
qc.x(1)
qc.z(0)
qc.cx(
    0, # Контроллирующий фотон
    1  # Фотон куда применяем
)

qc.draw('mpl')
````

Строка 3: $(|0> + |1>)|0>$

Строка 4: $(|0> + |1>)|1>$

Строка 5: $(|0> - |1>)|1>$

$= |01> - |11>$

Строка 6: $|01> - |10>$

**BB84** — первый протокол квантового распределения ключей, разработанный Беннеттом и Брассаром в 1984 году.

Представь, что Алиса хочет передать Бобу секретный ключ. Проблема: злая Ева может подслушивать канал.

***Основная идея BB84 на пальцах 🖐️***

1. **Алиса готовит фотоны:**
    - Берёт случайный бит: 0 или 1
    - Выбирает случайный "фильтр" (базис): прямой ➕ или диагональный ❌
    - Отправляет фотон Бобу

2. **Боб измеряет фотоны:**
    - Для каждого фотона случайно выбирает свой фильтр: ➕ или ❌
    - Измеряет фотон

    И биты, и базисы полностью случайные и независимые друг от друга!

    ```
    Алиса генерирует:
    Биты:   [1, 0, 1, 1, 0, ...]  ← случайные
    Базисы: [0, 1, 0, 1, 1, ...]  ← тоже случайные (НЕ зависят от битов!)

    Боб генерирует:
    Базисы: [0, 0, 1, 1, 0, ...]  ← свои случайные
    ```

    ***Важно: BB84 НЕ передаёт сообщение! Он создаёт общий секретный ключ!***

3. **Публичное сравнение:**
    - Алиса и Боб говорят по телефону: "У меня был фильтр ➕, ❌, ➕..."
    - НЕ говорят результаты! Только какой фильтр использовали
    - ***Оставляют только те биты, где фильтры совпали, остальные выбрасывают***

    ```
    Алиса отправила: [1, 0, 1, 1, 0, 1, 0, 1]
    Базисы Алисы:    [0, 1, 0, 1, 1, 0, 1, 0]
    Базисы Боба:     [0, 0, 1, 1, 0, 0, 1, 1]
                      ✓  ✗  ✗  ✓  ✗  ✓  ✓  ✗

    Оставляем только позиции с ✓:
    Общий ключ: [1, 1, 1, 0]
    ```

    Теперь у Алисы и Боба есть одинаковый секретный ключ `[1, 1, 1, 0]`.

    Затем они используют его для шифрования сообщений обычным симметричным шифром!

4. **Проверка на ~~Дусю~~ Еву:**
    - Жертвуют половиной битов: говорят друг другу эти биты
    - Если биты совпадают → канал чистый ✅
    - Если есть ошибки → Ева подслушивала! ⚠️

**Почему это работает?**

Если Ева измеряет фотон, она разрушает его квантовое состояние! Алиса и Боб увидят ошибки.

### [SLOP] Реализация в коде

````python
print("=" * 60)
print("         BB84 - КВАНТОВОЕ РАСПРЕДЕЛЕНИЕ КЛЮЧЕЙ")
print("=" * 60)

# ============================================================================
# Параметры
# ============================================================================
n = 20  # Количество фотонов
eve_present = True  # Есть ли подслушиватель?

# ============================================================================
# ШАГ 1: Алиса готовит фотоны
# ============================================================================
print("\n📤 ШАГ 1: Алиса готовит фотоны")
print("-" * 60)

alice_bits = [random.randint(0, 1) for _ in range(n)]
alice_bases = [random.randint(0, 1) for _ in range(n)]  # 0 = ➕, 1 = ❌

print(f"Биты Алисы:   {alice_bits}")
print(f"Базисы (0=➕, 1=❌): {alice_bases}")

# ============================================================================
# ШАГ 2: Передача (здесь может вмешаться Ева!)
# ============================================================================
print("\n📡 ШАГ 2: Передача фотонов")
print("-" * 60)

transmitted_bits = alice_bits.copy()

if eve_present:
    print("⚠️  Ева подслушивает!")
    eve_bases = [random.randint(0, 1) for _ in range(n)]

    # Ева измеряет фотоны (и разрушает их!)
    for i in range(n):
        if alice_bases[i] != eve_bases[i]:
            # Неправильный базис → случайный результат
            transmitted_bits[i] = random.randint(0, 1)
else:
    print("✅ Канал чистый (без Евы)")

# ============================================================================
# ШАГ 3: Боб измеряет
# ============================================================================
print("\n📥 ШАГ 3: Боб измеряет фотоны")
print("-" * 60)

bob_bases = [random.randint(0, 1) for _ in range(n)]
bob_results = []

for i in range(n):
    if alice_bases[i] == bob_bases[i]:
        # Базисы совпали → правильный результат
        bob_results.append(transmitted_bits[i])
    else:
        # Базисы не совпали → случайный результат
        bob_results.append(random.randint(0, 1))

print(f"Базисы Боба:  {bob_bases}")
print(f"Результаты:   {bob_results}")

# ============================================================================
# ШАГ 4: Публичное сравнение базисов (по телефону)
# ============================================================================
print("\n☎️  ШАГ 4: Алиса и Боб сравнивают базисы")
print("-" * 60)

sifted_alice = []
sifted_bob = []
matching_positions = []

for i in range(n):
    if alice_bases[i] == bob_bases[i]:
        sifted_alice.append(alice_bits[i])
        sifted_bob.append(bob_results[i])
        matching_positions.append(i)

print(f"Позиции с совпавшими базисами: {matching_positions}")
print(f"Просеянные биты Алисы: {sifted_alice}")
print(f"Просеянные биты Боба:  {sifted_bob}")
print(f"Всего битов после просеивания: {len(sifted_alice)}")

# ============================================================================
# ШАГ 5: Проверка на подслушивание
# ============================================================================
print("\n🔍 ШАГ 5: Проверка на Еву")
print("-" * 60)

if len(sifted_alice) == 0:
    print("❌ Недостаточно данных!")
else:
    # Жертвуем половиной битов для проверки
    test_size = len(sifted_alice) // 2

    errors = 0
    for i in range(test_size):
        if sifted_alice[i] != sifted_bob[i]:
            errors += 1

    error_rate = errors / test_size if test_size > 0 else 0

    print(f"Проверено битов: {test_size}")
    print(f"Ошибок: {errors}")
    print(f"Уровень ошибок: {error_rate * 100:.1f}%")

    # ШАГ 6: Финальный ключ
    print("\n🔑 ШАГ 6: Финальный ключ")
    print("-" * 60)

    if error_rate > 0.11:  # Порог для BB84
        print("⚠️  ОПАСНОСТЬ! Обнаружено подслушивание!")
        print("❌ Ключ отброшен. Протокол прерван.")
    else:
        # Оставшиеся биты = секретный ключ
        final_key_alice = sifted_alice[test_size:]
        final_key_bob = sifted_bob[test_size:]

        print(f"✅ Канал безопасен!")
        print(f"Финальный ключ Алисы: {final_key_alice}")
        print(f"Финальный ключ Боба:  {final_key_bob}")
        print(f"Длина ключа: {len(final_key_alice)} бит")

        if final_key_alice == final_key_bob:
            print("\n🎉 УСПЕХ! Ключи идентичны!")
        else:
            print("\n⚠️  Ключи не совпадают (возможны ошибки)")

print("\n" + "=" * 60)
print("Запустите снова с eve_present = True для проверки атаки!")
print("=" * 60)
````

## [SLOP] Протокол Эщкере E91

***E91 (Протокол Экерта) на пальцах 🖐️***

**Главное отличие от BB84:**
- BB84: Алиса посылает фотоны → Боб их ловит
- E91: Третий источник создаёт запутанные пары → Алиса и Боб получают по одному фотону из каждой пары

**Что такое запутанность?**
Представь две монетки-близнецы:
- Пока не смотришь — у каждой "и орёл, и решка" одновременно
- Как только смотришь на первую — она падает на орёл
- Мгновенно вторая тоже становится орлом! (или обе решки)
- Они связаны, даже если на разных концах Вселенной!

**Как работает E91:**

1. Источник создаёт запутанные пары:
    ```
    Источник: 👉 |Φ⁺⟩ = (|00⟩ + |11⟩)/√2
              /          \
        Алисе ←            → Бобу
    ```

2. Алиса и Боб случайно выбирают фильтры:
    ```
    Алиса выбирает из: [0°, 45°, 90°]
    Боб выбирает из:   [45°, 90°, 135°]
    ```

3. Измеряют свои фотоны:
    - Если фильтры одинаковые → результаты одинаковые (из-за запутанности!)
    - Если фильтры разные → проверяют корреляцию (для обнаружения Евы)

4. Сравнение:
    - Одинаковые углы → биты ключа
    - Разные углы → проверка через неравенство Белла

**Неравенство Белла как детектор Евы:**
Квантовая запутанность нарушает классическую логику!

1. Если кубиты запутаны (нет Евы):
    - Корреляция высокая: ~85% совпадений
    - S > 2 (параметр Белла)

2. Если Ева подслушивает:
    - Запутанность разрушается
    - Корреляция падает: ~50% (как случайность)
    - S ≤ 2

````python
print("=" * 70)
print("         E91 (ПРОТОКОЛ ЭКЕРТА) - КВАНТОВАЯ ЗАПУТАННОСТЬ")
print("=" * 70)

# ==========================================================================
# Параметры
# ==========================================================================
n_pairs = 40  # Количество запутанных пар
eve_present = True  # Подслушивает ли Ева?

# Углы измерения (в единицах по 45°)
# Алиса: 0°, 45°, 90°
# Боб:   45°, 90°, 135°
alice_angles = [0, 1, 2]  # 0=0°, 1=45°, 2=90°
bob_angles = [1, 2, 3]    # 1=45°, 2=90°, 3=135°

print(f"\nИспользуем {n_pairs} запутанных пар")
print(f"Алиса выбирает углы: {[f'{a*45}°' for a in alice_angles]}")
print(f"Боб выбирает углы:   {[f'{b*45}°' for b in bob_angles]}")

# ==========================================================================
# ШАГ 1: Источник создаёт запутанные пары
# ==========================================================================
print("\n" + "=" * 70)
print("1️⃣  Источник создаёт запутанные пары фотонов")
print("=" * 70)
print("Каждая пара в состоянии: |Φ⁺⟩ = (|00⟩ + |11⟩)/√2")
print("Это означает:")
print("  - Если Алиса получит 0, Боб тоже получит 0")
print("  - Если Алиса получит 1, Боб тоже получит 1")
print("  - НО! Пока не измерено - они в суперпозиции (и 0, и 1 одновременно)")
print("\n          Источник")
print("            🌟")
print("           /  \\")
print("    фотон /    \\ фотон")
print("         /      \\")
print("     Алиса      Боб")

# ==========================================================================
# ШАГ 2: Алиса и Боб выбирают углы и измеряют
# ==========================================================================
print("\n" + "=" * 70)
print("2️⃣  Алиса и Боб независимо выбирают углы и измеряют")
print("=" * 70)

alice_chosen_angles = []
bob_chosen_angles = []
alice_results = []
bob_results = []

for i in range(n_pairs):
    # Каждый случайно выбирает угол из своего набора
    alice_angle = random.choice(alice_angles)
    bob_angle = random.choice(bob_angles)

    alice_chosen_angles.append(alice_angle)
    bob_chosen_angles.append(bob_angle)

    # Симуляция квантовых измерений
    if eve_present:
        # Ева измеряет → разрушает запутанность → случайные результаты
        alice_result = random.randint(0, 1)
        bob_result = random.randint(0, 1)
    else:
        # Квантовая корреляция из-за запутанности
        base_result = random.randint(0, 1)

        if alice_angle == bob_angle:
            # Одинаковые углы → одинаковые результаты (100%)
            alice_result = base_result
            bob_result = base_result
        else:
            # Разные углы → корреляция зависит от разницы углов
            angle_diff = abs(alice_angle - bob_angle)

            # Квантовая механика: cos²(разница углов * 22.5°)
            # Упрощённо: чем ближе углы, тем выше корреляция
            if angle_diff == 1:  # 45° разница
                correlation = 0.85  # ~85% совпадений (квантовая магия!)
            else:  # 90° и больше
                correlation = 0.5   # ~50% совпадений

            alice_result = base_result
            bob_result = base_result if random.random() < correlation else (1 - base_result)

    alice_results.append(alice_result)
    bob_results.append(bob_result)

print(f"Алиса выбрала углы: {alice_chosen_angles[:10]}...")
print(f"Боб выбрал углы:    {bob_chosen_angles[:10]}...")
print(f"Результаты Алисы:   {alice_results[:10]}...")
print(f"Результаты Боба:    {bob_results[:10]}...")

# ==========================================================================
# ШАГ 3: Публичное сравнение углов
# ==========================================================================
print("\n" + "=" * 70)
print("3️⃣  Алиса и Боб публично сравнивают углы (НЕ результаты!)")
print("=" * 70)

# Биты ключа: где углы ОДИНАКОВЫЕ
key_bits_alice = []
key_bits_bob = []
key_positions = []

# Биты для проверки Белла: где углы РАЗНЫЕ
bell_check_data = []

for i in range(n_pairs):
    if alice_chosen_angles[i] == bob_chosen_angles[i]:
        # Одинаковые углы → для ключа
        key_bits_alice.append(alice_results[i])
        key_bits_bob.append(bob_results[i])
        key_positions.append(i)
    else:
        # Разные углы → для проверки на Еву
        bell_check_data.append({
            'alice_angle': alice_chosen_angles[i],
            'bob_angle': bob_chosen_angles[i],
            'alice_result': alice_results[i],
            'bob_result': bob_results[i]
        })

print(f"\nОдинаковые углы: {len(key_positions)} пар (для ключа)")
print(f"Разные углы:     {len(bell_check_data)} пар (для проверки Белла)")

# ==========================================================================
# ШАГ 4: Проверка на Еву через неравенство Белла
# ==========================================================================
print("\n" + "=" * 70)
print("4️⃣  Проверка на подслушивание через неравенство Белла")
print("=" * 70)

if len(bell_check_data) > 0:
    # Считаем корреляцию (сколько результатов совпали)
    matches = sum(1 for d in bell_check_data
                  if d['alice_result'] == d['bob_result'])
    correlation = matches / len(bell_check_data)

    # Упрощённая проверка Белла:
    # Квантовая запутанность даёт корреляцию ~85%
    # Классическая (Ева разрушила) даёт ~50%

    print(f"Проверено {len(bell_check_data)} пар с разными углами")
    print(f"Совпадений: {matches} из {len(bell_check_data)}")
    print(f"Корреляция: {correlation:.1%}")
    print()

    if correlation > 0.7:
        print("✅ Корреляция высокая (>70%) → Квантовая запутанность сохранена!")
        print("✅ Канал безопасен, Евы нет!")
        eve_detected = False
    else:
        print("⚠️  Корреляция низкая (<70%) → Запутанность разрушена!")
        print("⚠️  ПОДСЛУШИВАНИЕ ОБНАРУЖЕНО!")
        eve_detected = True
else:
    print("⚠️  Недостаточно данных для проверки Белла")
    eve_detected = True

# ==========================================================================
# ШАГ 5: Финальный ключ
# ==========================================================================
print("\n" + "=" * 70)
print("5️⃣  Создание финального ключа")
print("=" * 70)

if eve_detected:
    print("❌ Протокол прерван из-за подслушивания!")
else:
    # Жертвуем половиной битов для дополнительной проверки
    test_size = len(key_bits_alice) // 2

    errors = sum(1 for i in range(test_size)
                 if key_bits_alice[i] != key_bits_bob[i])

    print(f"\nДополнительная проверка (жертвуем {test_size} битов):")
    print(f"Ошибок: {errors}/{test_size}")

    # Финальный ключ
    final_key = key_bits_alice[test_size:]

    print(f"\n🔑 ФИНАЛЬНЫЙ СЕКРЕТНЫЙ КЛЮЧ:")
    print(f"   Ключ: {final_key}")
    print(f"   Длина: {len(final_key)} бит")
    print(f"   ✅ Ключи Алисы и Боба идентичны!")
````

## [SLOP] Сверхплотное кодирование

~~Карается 20.3 КоАП РФ, а при рецидивах и 282.4 УК РФ~~

**Сверхплотное кодирование (Superdense Coding)** — протокол передачи 2 классических битов с помощью 1 кубита, используя предварительно распределённую запутанность.

**Идея:**

1. Алиса и Боб заранее делают пару запутанных кубитов — второй из них уже у Боба, а первый Алиса ему вскоре отправит
2. Алиса кодирует 2 бита на первом кубите локальными операциями
3. Алиса отправляет этот кубит Бобу
4. Боб декодирует оба бита, измерив оба кубита вместе

Таким образом, боб получает 2 бита информации при помощи извлечения оной из всего 1 кубита! Да, по факту для этого нужна ПАРА запутанных кубитов (и даже если Боб будет создавать эти пары у себя на хате, ему надо будет один из них отправить Алисе), но всё же!

**Кодирование 2 битов:**

1. $00 → I → |Φ⁺⟩$
2. $01 → X → |Ψ⁺⟩$
3. $10 → Z → |Φ⁻⟩$
4. $11 → XZ → |Ψ⁻⟩$

````python
# 00 (=0), 01 (=1), 10 (=2) либо 11 (=3)
DIGIT_TO_ENCODE = 3
````

### Версия с семинара:

````python
qc = QuantumCircuit(
    4, # 4 кубита
    2  # 2 классических бита (для измерения в конце)
)

# Кодируем сообщение первыми двумя кубитами
if (DIGIT_TO_ENCODE & 0b01):
    qc.x(0)
if (DIGIT_TO_ENCODE & 0b10):
    qc.x(1)
qc.barrier()

# Вводим кубит Боба и транспортный кубит в состояние КЗ ("ЦЕПЬ БЕЛЛА")
qc.h(2)
qc.cx(2, 3)
qc.barrier()

# Алиса передаёт транспортный кубит Бобу
qc.cx(1, 2)
qc.cz(0, 2)
qc.barrier()

# Боб расшифровывает переданный транспортный кубит ("ОБРАТНАЯ ЦЕПЬ БЕЛЛА")
qc.cx(2, 3)
qc.h(2)
qc.barrier()

# Вуаля, всего одним кубитом мы передали число длиной 2 классических бита!
qc.measure(2, 0)
qc.measure(3, 1)

qc.draw("mpl")
````

````python
simulator = AerSimulator()

compiled_circuit = transpile(qc, simulator)

job = simulator.run(compiled_circuit, shots=1024)
result = job.result()
counts = result.get_counts(compiled_circuit)

print("Результат измерения:", counts)
````

## Квантовая телепортация

_Штош, на семинаре нам НИХУЯ нормально не объяснили, т.ч. тут реализация по книжке!_

_Полный текст на стр. 170, чуть выше объясняется сверхполтное кодлирование._

Алиса и Боб находятся далеко друг от друга. У каждого имеется один электрон. Электроны находятся в запутанном состоянии. У Алисы имеется также еще один электрон. Он находится в состоянии $a| \ 0 > + \ b| \ 1 >$. Алиса ничего не знает об амплитудах вероятности $a$ и $b$, но она и Боб хотят изменить электрон Боба так, чтобы тот получил состояние $a| \ 0 > + \ b| \ 1 >$, как у Алисы.

Они хотят телепортировать Бобу состояние электрона Алисы, то есть, грубо говоря, хотят, чтобы у Боба получились те же самые амплитуды вероятностей $a$ и $b$.

Самый простой способ сделать это — послать два кубита Алисы в вентиль управляемого НЕ. Это станет первым шагом. Вторым шагом станет применение вентиля Адамара к верхнему кубиту. Фактически Алиса должна подать два своих кубита на вход обратной цепи Белла.

Теперь Алиса должна измерить два своих электрона в стандартном базисе. Она получит одно из состояний — $| \ 00 >$, $| \ 01 >$, $| \ 10 >$ или $| \ 11 >$, — причем каждое с вероятностью $\frac14$.

Чтобы прийти к окончательному итогу, Алиса должна сообщить Бобу, в какой из четырех возможных ситуаций он находится. Для этого она должна послать Бобу два **классических** бита информации — `00, 01, 10 или 11,` — соответствующих результатам ее измерений. Эти биты можно отправить в любом виде, например в виде сообщения В МЕССЕНДЖЕРЕ МАКС.

- Если Боб получит `00`, он будет знать, что его кубит уже имеет нужное состояние и ничего делать с ним не требуется.
- Если Боб получит `01`, он будет знать, что его кубит имеет состояние $a| \ 1 > + \ b| \ 0 >$ и ему нужно применить вентиль X.
- Если Боб получит `10`, он будет знать, что его кубит имеет состояние $a| \ 0 > - \ b| \ 1 >$ и ему нужно применить вентиль Z.
- Если Боб получит `11`, он будет знать, что его кубит имеет состояние $a| \ 1 > - \ b| \ 0 >$ и ему нужно применить вентили X и Z.

В каждом случае Боб получает кубит в состоянии $a| \ 0 > + \ b| \ 1 >$, то есть в состоянии исходного кубита, которое Алиса хотела телепортировать.

### Реализация в коде:

````python
qc = QuantumCircuit(
    3, # 3 кубита
    2  # 2 классических бита
)

# Инициализируем какое-нибудь состояние кубита Алисы
qc.initialize([
    np.sqrt(0.7),
    np.sqrt(0.3),
], [0])

# Сперва, запутываем кубиты Алисы и Боба
qc.h(1)
qc.cx(1, 2)

# Шаг 1 - делаем CNOT для кубита Алисы от неизвестного кубита
qc.cx(0, 1)

# Шаг 2 - применяем вентиль Адамара к неизвестному кубиту
qc.h(0)

# Алиса измеряет свой кубит и смотрим, чему равно его состояние. Таким образом
# она "разрушает" своё состояние, и, в силу запутанности, оно переходит к Бобу
qc.measure(0, 0)
qc.measure(1, 1)

# Алиса отправила информацию о состоянии кубитов Бобу В МЕССЕНДЖЕРЕ МАКС,
# теперь, на основе этой инфомации, Боб применяет соответствуюбщие вентили
with qc.if_test((qc.clbits[1], 1)):
    qc.x(2) # Если второй бит = 1, применить X
with qc.if_test((qc.clbits[0], 1)):
    qc.z(2) # Если второй бит = 1, применить Z

# Важно помнить, что амплитуды a, b никогда не существуют локально у Боба
# до коррекции, они закодированы в совместном состоянии + результате измерения!
qc.draw('mpl')
````

Проверим, что "телепортированный" кубит имеет те же вероятности, что и изначальный:

````python
stats = count_stats(qc).items()

a_shots = 0; b_shots = 0

for (qbits, count) in stats:
    if qbits[0] == '0':
        a_shots += count
    else:
        b_shots += count

total_shots = a_shots + b_shots

print(f"P(a) = {a_shots / total_shots}, P(b) = {b_shots / total_shots}")
````

**Алгоритм Гровера** — квантовый алгоритм поиска в неструктурированной базе данных за O(√N) вместо классических O(N).

**Компоненты:**
  - Оракул — помечает правильный ответ
  - Диффузия (инверсия относительно среднего) — усиливает амплитуду правильного ответа
