```python
def caesar(text, shift):
    result = ""
    for char in text:
        if char.isalpha():
            start = ord('А') if char.isupper() else ord('а')
            result += chr((ord(char) - start + shift) % 32 + start)
        else:
            result += char
    return result
```

```python
text = re.sub(r'[^а-яё]', '', super_cool_message)

def letter_frequency(text):
    return Counter(text)

def ngram_frequency(text, n):
    return Counter(text[i:i+n] for i in range(len(text) - n + 1))

letters = letter_frequency(text)

def print_top(counter, top=20):
    for k, v in counter.most_common(top):
        print(f"{k}: {v}")

print('Буквы по частоте:')
print_top(letters)
```
